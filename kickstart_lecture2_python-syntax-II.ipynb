## Control Structures: If... else...

```python
# Define a variable to check against in the conditional structure
check_var = 45

# if

check_var = 45
```

```python
# Conditional statement to check if check_var equals 45
# If true, increment check_var by 10, calculate its square, and print the result
if check_var == 45:
    check_var += 10
    check_var_square = check_var ** 2
    print(check_var_square)
```

```python
# Another example of conditional structure with an else clause
check_var = 45
if check_var == 45:
    check_var += 10
    check_var_square = check_var ** 2
    print(check_var_square)
else:
    print('check_var does not equal 45')
```

## Control Structures: For and While Loops

```python
# Using a "while" loop to print powers of 2 for each j in range(10)
i = 0
while i<10:
    result = 2**i
    print(i, result)
    i+=1
```

```python
# The same with a "for" loop
for j in range(10):
    result = 2**j
    print(j,result)
```

## Functions

```python
# Function that calculates and prints the cube of the number 5
def cube_five_function():
    number = 5
    cube_of_number = number ** 3
    print('Cube of 5 is:', cube_of_number)
```

```python
# Call the cube_five_function to display the cube of 5
cube_five_function()
```

```python
# Function that accepts a number as input and prints its cube
def cube_a_number_function(number):
    cube_of_number = number ** 3
    print(f'Cube of {number} is:', cube_of_number)
```

```python
# Call the cube_a_number_function with 100 as input
cube_a_number_function(100)
```

```python
# Capture the result of cube_a_number_function with 100 as input 
# but note it returns None
res = cube_a_number_function(100)
```

```python
# Display the value of res (will be None) 
# since cube_a_number_function does not return a value
res
```

```python
# Add "return" values
def cube_a_number_function_advanced(number):
    cube_of_number =number**3
    print(cube_of_number)
    return cube_of_number
```

```python
# Call the advanced cube function and store the result
res2 = cube_a_number_function_advanced(100)
res2
```

```python
# Attempt to call the advanced cube function without arguments,
# expected to raise an error
cube_a_number_function_advanced()
```

```python
# Assign defaults
def cube_a_number_function_advanced2(number=1):
    cube_of_number =number**3
    print(cube_of_number)
    return cube_of_number
```

```python
# Call the advanced cube function with a default argument, calculating the cube of 5
cube_a_number_function_advanced2(5)
```
