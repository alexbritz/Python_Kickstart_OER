### Array Handling: Concatenating and Reshaping

```python
# First create 3 1D-arrays
a1 = np.array([1,2,3])
a2 = np.array([4,5,6])
a3 = np.array([7,8,9])
```

```python
# And merge them 
array_merged = np.concatenate([a1,a2,a3])
array_merged
```

```python
# Test reshape
array_2D_from_1D = array_merged.reshape((3,3))
array_2D_from_1D
```

```python
# Stacking: vertical and horizontal
array_vstack = np.vstack((a1,a2,a3))
array_vstack
```

```python
np.hstack((a1.reshape(3,1),a2.reshape(3,1),a3.reshape(3,1)))
```

```python
# Transpose
array_vstack.transpose()
```

```python
# Check available methods with dir
dir(array_1D)
```

### Math Operations

```python
# Element-wise multiplication (use 3 1D-arrays from above)
a1,a2,a3
```

```python
a1*a2
```

```python
# Flipping row to colum and multiply
a1.reshape(-1,1)*a2
```

```python
# The same works for division
a1.reshape(-1,1)/a2
```

```python
a1**a2
```
