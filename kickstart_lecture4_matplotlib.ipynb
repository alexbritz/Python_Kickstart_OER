## Python Kickstart Lecture 4: Matplotlib for Plotting

```python
# Import libraries: Numpy and Matplotlib

import numpy as np
import matplotlib.pyplot as plt
```

### Line Plot

```python
# Generate random x-y Data

x = np.linspace(-10,10,201)
y = x**3-1
```

```python
x2 = np.linspace(-10,10,21)
y2 = x2**3-1
```

```python
# Create plot

plt.figure()
plt.plot(x,y, ':', color = 'red')
plt.plot(x2,y2, 'x',color = 'blue')
```

```python
# Add axis labels, legends, fontsize

plt.figure()
plt.plot(x,y, 'r:', label = 'quasi-continous')
plt.plot(x2,y2, 'bx',label = 'discrete')
plt.legend(fontsize = 14)
plt.xlabel('X', fontsize = 14)
plt.ylabel('Y = X$^3$ - 1', fontsize = 14)
```

### Histogram

```python
# Generate gaussian distributed data

gaussian_data = 3 + 2.5 * np.random.randn(10000)
```

```python
plt.figure()
plt.hist(gaussian_data, 50)
plt.xlabel('x')
plt.ylabel('occur')
```

### Loading Data and Visualizing

```python
# Load XAS data with loadtxt

data = np.loadtxt('xas_FeCO5_from_materialsproject.tsv',
          delimiter='\t',
          skiprows = 4,
          usecols = [0,1]
          )
x_data = data[:,0]
y_data = data[:,1]
```

```python
# Create a plot
plt.figure()
plt.plot(x_data,y_data)

plt.savefig('Fe_xas.png', dpi = 300) # pixelated
plt.savefig('Fe_xas.eps') # vectorized
```
