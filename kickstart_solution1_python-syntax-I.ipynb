```python
# a) Create an integer and a floating point variable.

int_var = 10
float_var = 10.5

print(int_var)
print(float_var)
```

```python
# Multiply both variables from a) with each other and assign the result to a new variable.
#  Which types is this new variable? Use the type() function.  

result = int_var * float_var
print(result) # This will print the result of the multiplication
print(type(result)) # This will print the type of the result
```

```python
# c) Create a string variable. Check how many characters the 
# string variable has using the len() function.
#  Multiply this string variable by 5. What do you get?  

string_var = "Hello"
print(len(string_var)) # This will print the length of the string
```

```python
print(string_var * 5) # This will print the string repeated 5 times
```

```python
# d) Create a list which should be filled with 10 elements of mixed types.  

mixed_list = [1, 'Hello', 3.14, True, 'World', [1, 2], (3, 4), {5: "five"}, {6, 7}, complex(1, 2)]
mixed_list
```

```python
# e) Access the fifth element of the list and check its type.
print(mixed_list[4])
print(type(mixed_list[4]))
```

```python
# f) Access the last three elements of the list. 
# BONUS: Can you access all elements with even indices?  
print(mixed_list[-3:]) # This will print the last three elements of the list
```

```python
mixed_list[::2] # This will print all elements with even indices
```

```python
# BONUS: Use the dir() command to check which methods are 
# available for objects of type list. Test three different methods!

dir(mixed_list)
```

```python
# Append a new element to the list and print the updated list (actually, three new elements :) ).
print(mixed_list) # This will print the original list

mixed_list.append("New Element") # Appending a new element
mixed_list.append(42) # Appending an integer
mixed_list.append([1, 2, 3]) # Appending a list

print(mixed_list) # This will print the updated list
```

```python
# Remove the string "Hello" from the list and print the updated list.
mixed_list.remove("Hello") # Removing the string "Hello"
print(mixed_list) # This will print the updated list
```

```python
# Reverse the order of elements in the list and print it.
mixed_list.reverse() # Reversing the list
```
