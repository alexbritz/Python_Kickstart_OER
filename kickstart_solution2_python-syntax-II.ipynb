```python
# a) i) Create two variables and assign string values to them. 
# One string should be shorter than 10 characters, and the other should be longer.  

short_string = 'Hello'
long_string = 'This is a longer string than the first one.'
```

```python
# a) ii) Write a control structure that checks whether each string is shorter than
#  or equal to 10 characters or longer. The output should be a statement like:  
# "The string is longer than 10 characters"  
# "The string is up to 10 characters long"  

string_to_check = short_string # You can change this to long_string to check the other string
if len(string_to_check) <= 10:
    print('The string is up to 10 characters long')
else:
    print('The string is longer than 10 characters')
```

```python
# a) Modify the control structure so that the printed statement includes the actual string, like:  
# "The string '[your string]' is longer than 10 characters"  
# "The string '[your string]' is up to 10 characters long" 

string_to_check = long_string # You can change this to long_string to check the other string
if len(string_to_check) <= 10:
    print('The string "' +string_to_check+'" is up to 10 characters long')
else:
    print('The string "' +string_to_check+'" is longer than 10 characters')
```

```python
# b) i) Write a function that calculates and returns the value of a third-order polynomial:
#  ax^3+ bx^2 + cx + d.. Choose coefficients a,b,c,d test the function with a value for x. 

def polynomial_3rd_order(x):
    return 1.3*x**3 + 4.5*x**2 + 0.4*x + 9.1

# test the function with a value for x
x = 5
result = polynomial_3rd_order(x)
print(result)
```

```python
# ii)Make the polynomial function more general. Write another version of 
# the function that takes both x x and the coefficients a, b, c, d as input arguments. 

def polynomial_3rd_order_general(a, b, c, d, x):
    return a*x**3 + b*x**2 + c*x + d

# test the function with a value for x
a = 1
b = 2
c = 3
d = 4
x = 5
result = polynomial_3rd_order_general(a, b, c, d, x)
print(result)

```

```python
# b) iii) Transform the control structure code from part (a) into a function. 
# The function should take a string as input and print the corresponding statement about its length. 

def check_string_length(string_to_check):
    if len(string_to_check) <= 10:
        statement = 'The string "' +string_to_check+'" is up to 10 characters long'
    else:
        statement = 'The string "' +string_to_check+'" is longer than 10 characters'
    print(statement)
    return statement

# test the function with both strings
check_string_length(short_string)
check_string_length(long_string)
```
