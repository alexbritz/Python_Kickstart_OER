# Exercise 4: Matplotlib
a) Create a numpy array with x-values from -10 to 10 and a step size that you like. Use this array with x-values to create y-values of the polynomial y = 2*x**3 + 2.3*x -5.

b) Plot the polynomial using matplotlib.

c) Add a legend.

d) Add axis labels.

```python
# Import the numpy and matplotlib library
import numpy as np
import matplotlib.pyplot as plt
```

```python
# a) Create a numpy array with x-values from -10 to 10 and a step size that you like. 
# Use this array with x-values to create y-values of the polynomial y = 2*x**3 + 2.3*x -5.a

# Create a numpy array with x-values from -10 to 10
x = np.arange(-10, 10, 0.1)  # Adjust the step size as needed

# Calculate the y-values using the polynomial equation
y = 2 * x**3 + 2.3 * x - 5
```

```python
# b) Plot the polynomial using matplotlib.
plt.figure()
plt.plot(x, y)
```

```python
# c) Add a legend.

# Recreate the plot with a legend
plt.figure()
plt.plot(x, y, label='y = 2*x^3 + 2.3*x - 5')
plt.legend()
```

```python
# d) Add axis labels.

# Recreate the plot with a legend and axis labels
plt.figure()
plt.plot(x, y, label='y = 2*x^3 + 2.3*x - 5')
plt.legend()

# Add axis labels
plt.xlabel('x')
plt.ylabel('y')
```
